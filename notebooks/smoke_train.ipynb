# Smoke training check

The images written by `cloudie.make_smoke_data` are synthetic. They are not cyclones. A fit on them only shows that the training loop runs. Do not report scores from this notebook. Cloudie does not issue forecasts or warnings.


```python
from pathlib import Path

from cloudie.config import load_config
from cloudie.evaluate import evaluate
from cloudie.make_smoke_data import write_smoke
from cloudie.train import train

config = load_config("config/config_smoke.yaml")
manifest = write_smoke(Path("data/smoke"), size=int(config["data"]["image_size"]))
checkpoint = train(config, manifest)
result = evaluate(config, checkpoint)
print(result.note)
print("present_as_forecast", result.present_as_forecast)
```
